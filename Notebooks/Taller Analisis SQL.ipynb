```python

%sql
SELECT 
    pais,
    pib_billones_usd,
    ROUND(pib_billones_usd, 1) AS pib_redondeado,
    CEIL(pib_billones_usd) AS pib_techo,
    FLOOR(pib_billones_usd) AS pib_piso,
    TRUNCATE(pib_billones_usd, 0) AS pib_truncado
FROM workspace.electiva.indicadores_paises;
```

```python

%sql
-- Ejemplo: Calcular la desviación absoluta de la inflación respecto a la meta del banco central (3.0%)
SELECT 
    pais,
    tasa_inflacion_pct,
    ABS(tasa_inflacion_pct - 3.00) AS desviacion_meta_inflacion
FROM workspace.electiva.indicadores_paises;
```

```python

%sql
SELECT 
    pais,
    pib_billones_usd,
    LN(pib_billones_usd) AS log_pib, -- Transformación logarítmica econométrica
    POWER(1 + (tasa_inflacion_pct / 100), 2) - 1 AS inflacion_acumulada_2anios,
    LOG10(pib_billones_usd) AS log_pib_base10, --Logaritmo en base 10 del pib_billones_usd,
    EXP(pib_billones_usd) AS exp_pib, --exponencial del pib_billones_usd
    sqrt(pib_billones_usd) AS sqrt_pib --raiz cuadrada del pib_billones_usd
FROM  workspace.electiva.indicadores_paises;
```

```python

%sql
SELECT 
    COUNT(*) AS total_paises_analizados,
    COUNT(subsidios_millones) AS paises_con_subsidios_registrados,
    COUNT(DISTINCT region_economica) AS bloques_economicos,
    SUM(pib_billones_usd) AS pib_mundial_acumulado,
    ROUND(AVG(tasa_inflacion_pct), 2) AS inflacion_promedio_global,
    MIN(tasa_inflacion_pct) AS menor_inflacion,
    MAX(tasa_inflacion_pct) AS mayor_inflacion,
    ROUND(STDDEV(tasa_inflacion_pct), 2) AS volatilidad_inflacion
FROM workspace.electiva.indicadores_paises;
```

```python

%sql
SELECT 
    region_economica,
    COUNT(*) AS cantidad_paises,
    ROUND(SUM(pib_billones_usd), 2) AS pib_total_region,
    ROUND(AVG(CAST(tasa_desempleo_pct AS DECIMAL(5,2))), 2) AS desempleo_promedio
FROM workspace.electiva.indicadores_paises
WHERE poblacion_millones >= 50 -- 1. Filtra economías con al menos 50 millones de habitantes
GROUP BY region_economica       -- 2. Agrupa los datos por Región Económica
HAVING AVG(tasa_inflacion_pct) > 3.00 -- 3. Filtra regiones cuya inflación promedio supere el 3%
ORDER BY pib_total_region DESC;
```

```python
%sql
SELECT 
    pais,
    ROUND((pib_billones_usd * 1000000000) / (poblacion_millones * 1000000), 2) AS pib_per_capita_usd,
    COALESCE(CAST(subsidios_millones AS DECIMAL(10,2)), 0) AS subsidios_reportados
FROM workspace.electiva.indicadores_paises;
```

Respuesta del punto 1: Al observar los resultados generados por este analisis, lo que obtenemos es una radiografía económica mucho más realista y comparable de los países, gracias a dos transformaciones clave:

Estandarización de la Riqueza (PIB per cápita):
Al escalar los valores de billones y millones para calcular el PIB per cápita exacto en dólares, eliminamos el sesgo del tamaño poblacional. Ya que esto nos permite comparar de forma justa, economías totalmente distintas. Por ejemplo, en los resultados vemos que el tamaño de la economía en bruto ya no importa tanto, sino la productividad por habitante; esto visibiliza la verdadera calidad de vida y el poder adquisitivo promedio en cada país, permitiendo comparar naciones pequeñas pero ricas frente a gigantes demográficos.

Visibilidad de la Política Fiscal (Manejo de Subsidios):
El tratamiento de los datos de la columna de subsidios, es fundamental para el análisis estadístico. En la economía real, la ausencia de un dato (La presencia de 1 NULL). Esto no significa que no sepamos qué pasó, sino que muchas veces el país no destinó recursos a subsidios o hay un vacío en la transparencia fiscal. Al forzar estos nulos a "0.00" con el COALESCE y estandarizarlos como decimales, logramos limpiar la base de datos. Ahora se puede ver gráficamente qué estados intervienen en su economía y cuáles operan con un gasto social de cero, sin que los datos vacíos nos generen errores al momento de calcular promedios globales o regionales.

```python
%sql
SELECT 
    region_economica,
    SUM(poblacion_millones) AS poblacion_total,
    MAX(CAST(tasa_desempleo_pct AS DECIMAL(5,2))) AS max_desempleo,
    MIN(CAST(tasa_desempleo_pct AS DECIMAL(5,2))) AS min_desempleo,
    ROUND(AVG(tasa_inflacion_pct), 2) AS promedio_inflacion
FROM workspace.electiva.indicadores_paises
GROUP BY region_economica
HAVING SUM(pib_billones_usd) >= 3.0;
```

Respuesta del punto 2: Al analizar este comparativo macroeconómico regional filtrado solo para los bloques con mayor peso global como (PIB acumulado superior a 3.0 billones de dólares), puedo extraer conclusiones sobre la estabilidad y los retos de cada región:

Estabilidad vs. Volatilidad en el Mercado Laboral:
Si miramos las brechas de desempleo, Norteamérica presenta el mercado laboral más homogéneo y estable de la muestra, con una banda muy estrecha (mínimo de 3.7% y máximo de 5.8%). En contraste, Medio Oriente / MENA y Asia-Pacífico muestran desigualdades internas extremas. Medio Oriente tiene la brecha más dramática: un país prácticamente en pleno empleo (0.1% de desempleo) conviviendo en la misma región con economías colapsadas que alcanzan un 22.3% de desocupación. Esto nos dice que agrupar por región no siempre significa homogeneidad económica.

El Reto Demográfico Asiático:
El peso de Asia-Pacífico es indiscutible. Ya que con más de 4,288 millones de habitantes, esta región concentra la inmensa mayoría de la fuerza laboral mundial. Sin embargo, logran mantener una inflación promedio bastante controlada (5.79%), muy alineada con los niveles de Europa (5.92%), lo que demuestra una alta capacidad de absorción y producción a gran escala frente al resto de continentes.

La Presión Inflacionaria en Economías Emergentes:
Finalmente, los datos de inflación promedio revelan quiénes están sufriendo más el costo de vida. Mientras Norteamérica y Europa mantienen promedios de un dígito (3.00% y 5.92% respectivamente), Latinoamérica (LATAM) y Medio Oriente / MENA enfrentan choques severos, promediando 11.58% y 23.16% respectivamente. Esto evidencia que los bloques en vías de desarrollo de nuestra muestra enfrentan debilidades monetarias y crisis cambiarias que impactan fuertemente el poder adquisitivo de sus poblaciones "como se ve reflejado en los casi 600 millones de habitantes de LATAM".

En conclusión, este analisis demuestra que superar la barrera de los 3.0 billones de dólares en PIB no garantiza estabilidad interna. Las potencias consolidadas muestran indicadores compactos, mientras que los bloques emergentes operan con altísimas brechas internas de inflación y desempleo.
